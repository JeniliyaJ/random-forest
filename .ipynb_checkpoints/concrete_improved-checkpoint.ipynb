# Concrete Strength Prediction - Corrected & Improved Pipeline

Order of work:
1. Load + duplicates check
2. Feature engineering (row-wise, so no leakage)
3. Proper split: group-wise (same mix never in both train & test)
4. Baseline vs improved models in `Pipeline` (no leakage)
5. Tuning
6. Final test on hold-out
7. Feature importance
8. Prediction interval
9. Business: minimum-cost mix finder
10. Save model

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns, joblib
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, RidgeCV
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor, GradientBoostingRegressor
from sklearn.model_selection import GroupKFold, GroupShuffleSplit, cross_validate, GridSearchCV
from sklearn.inspection import permutation_importance
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

SEED = 42
```

## Step 1: Load data + duplicates

```python
df = pd.read_csv('concrete.csv')
print('shape:', df.shape)
print('missing:', df.isna().sum().sum())
print('duplicate rows:', df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)   # duplicates train & test-la irundha score fake-ah erum
print('after drop:', df.shape)
```

## Step 2: Feature engineering
Ellam row-wise calculation, so split-ku munnadi pannalam (leakage illa).
- `w_c_ratio` = water / cement (unga `cement_water_ratio` thaan, correct peru)
- `w_binder` = water / (cement + slag + ash)  <- concrete-la romba important
- `log_age` = strength age-oda log curve follow pannum
- Skew handling-ku PowerTransformer venam, log_age enough. Tree models-ku skew prechanai illa.

```python
def add_features(d):
    d = d.copy()
    d['binder']     = d['cement'] + d['slag'] + d['ash']
    d['w_c_ratio']  = d['water'] / d['cement']
    d['w_binder']   = d['water'] / d['binder']
    d['sp_binder']  = d['superplastic'] / d['binder']
    d['agg_ratio']  = d['coarseagg'] / d['fineagg']
    d['log_age']    = np.log(d['age'])
    return d

df = add_features(df)
TARGET = 'strength'
MIX_COLS = ['cement','slag','ash','water','superplastic','coarseagg','fineagg']  # age thavira
FEATURES = [c for c in df.columns if c != TARGET]
X, y = df[FEATURES], df[TARGET]
FEATURES
```

## Step 3: Group-wise split
Same mix-ai pala ages-la test pannirukkanga. Random split-la oru mix train-um test-um vandhudum (leakage).
Group = 7 ingredient values. Oru mix-oda ella ages-um onnaa train-la or test-la irukkum.

```python
groups = df[MIX_COLS].round(1).astype(str).agg('-'.join, axis=1)
print('unique mixes:', groups.nunique(), 'of', len(df), 'rows')

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=SEED)
train_idx, test_idx = next(gss.split(X, y, groups))
X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
g_train = groups.iloc[train_idx]
print(len(X_train), len(X_test))
```

## Step 4: Models in Pipelines + group cross-validation
Ella scaling/fit-um pipeline ulle, so CV-la ovvoru fold-um train data-la mattum fit aagum.
Baseline = unga original linear (raw features). Appuram improved models.

```python
RAW = MIX_COLS + ['age']                       # original features
LIN = MIX_COLS + ['log_age', 'w_binder']       # linear-ku nalla features

def sel(cols):
    from sklearn.compose import ColumnTransformer
    return ColumnTransformer([('keep', 'passthrough', cols)])

models = {
  'Linear (raw)':        Pipeline([('s', sel(RAW)), ('sc', StandardScaler()), ('m', LinearRegression())]),
  'Linear (log_age+w/b)':Pipeline([('s', sel(LIN)), ('sc', StandardScaler()), ('m', LinearRegression())]),
  'Poly2 + Ridge':       Pipeline([('s', sel(LIN)), ('sc', StandardScaler()),
                                   ('p', PolynomialFeatures(2, include_bias=False)),
                                   ('sc2', StandardScaler()), ('m', RidgeCV(alphas=np.logspace(-2, 3, 20)))]),
  'Random Forest':       Pipeline([('m', RandomForestRegressor(n_estimators=300, min_samples_leaf=2, random_state=SEED, n_jobs=-1))]),
  'HistGradBoost':       Pipeline([('m', HistGradientBoostingRegressor(learning_rate=0.05, max_iter=500, random_state=SEED))]),
}

cv = GroupKFold(n_splits=5)
scoring = {'r2': 'r2', 'rmse': 'neg_root_mean_squared_error', 'mae': 'neg_mean_absolute_error'}
rows = []
for name, pipe in models.items():
    res = cross_validate(pipe, X_train, y_train, groups=g_train, cv=cv, scoring=scoring)
    rows.append({'model': name,
                 'R2':   f"{res['test_r2'].mean():.3f} +- {res['test_r2'].std():.3f}",
                 'RMSE': f"{-res['test_rmse'].mean():.2f} +- {res['test_rmse'].std():.2f}",
                 'MAE':  f"{-res['test_mae'].mean():.2f} +- {res['test_mae'].std():.2f}"})
pd.DataFrame(rows)
```

## Step 5: Hyperparameter tuning (best model-ku)
`max_depth=50/60` waste. Idhu useful params: `learning_rate`, `max_leaf_nodes`, `min_samples_leaf`, `l2_regularization`.

```python
param_grid = {
    'm__learning_rate':     [0.03, 0.05, 0.1],
    'm__max_leaf_nodes':    [8, 15, 31],
    'm__min_samples_leaf':  [5, 10, 20],
    'm__l2_regularization': [0.0, 1.0],
}
base = Pipeline([('m', HistGradientBoostingRegressor(max_iter=500, random_state=SEED))])
grid = GridSearchCV(base, param_grid, cv=GroupKFold(5), scoring='neg_root_mean_squared_error', n_jobs=-1)
grid.fit(X_train, y_train, groups=g_train)
print('best params:', grid.best_params_)
print('CV RMSE:', -grid.best_score_)
best_model = grid.best_estimator_
```

## Step 6: Final evaluation (hold-out test, oru dhadavai mattum)

```python
def report(name, model):
    p = model.predict(X_test)
    print(f"{name:22s} R2={r2_score(y_test,p):.3f}  RMSE={np.sqrt(mean_squared_error(y_test,p)):.2f}  MAE={mean_absolute_error(y_test,p):.2f}")
    return p

fitted = {}
for name in ['Linear (raw)', 'Poly2 + Ridge', 'Random Forest']:
    models[name].fit(X_train, y_train); fitted[name] = models[name]
    report(name, models[name])
pred = report('Tuned HistGradBoost', best_model)

plt.figure(figsize=(5,5))
plt.scatter(y_test, pred, alpha=.6)
lim = [y_test.min(), y_test.max()]
plt.plot(lim, lim, 'r--'); plt.xlabel('Actual (MPa)'); plt.ylabel('Predicted (MPa)'); plt.title('Actual vs Predicted')
plt.show()
```

## Step 7: Feature importance (permutation - impurity importance-ai vida reliable)

```python
pi = permutation_importance(best_model, X_test, y_test, n_repeats=20, random_state=SEED, scoring='r2')
imp = pd.Series(pi.importances_mean, index=X_test.columns).sort_values()
imp.plot.barh(figsize=(6,5), title='Permutation importance (drop in R2)')
plt.show()
```

## Step 8: Prediction interval (90%)
Business-la "predicted 42 MPa" vida "42 MPa, 90% sure 37-47 kulla" nu solradhu useful.
Quantile gradient boosting use pannrom.

```python
q_lo = GradientBoostingRegressor(loss='quantile', alpha=0.05, n_estimators=300, max_depth=3, learning_rate=0.05, random_state=SEED).fit(X_train, y_train)
q_hi = GradientBoostingRegressor(loss='quantile', alpha=0.95, n_estimators=300, max_depth=3, learning_rate=0.05, random_state=SEED).fit(X_train, y_train)

lo, hi = q_lo.predict(X_test), q_hi.predict(X_test)
coverage = np.mean((y_test >= lo) & (y_test <= hi))
print(f'90% interval coverage on test: {coverage:.2%}  (target ~90%)')
print(f'avg interval width: {np.mean(hi-lo):.1f} MPa')
```

## Step 9: Business use - minimum cost mix finder
Target strength-ku (e.g. 40 MPa @ 28 days) cheapest mix-ai search pannum.
**PRICES placeholder. Unga company-oda real price/kg podunga.**
Safe-ah irukka, prediction-oda *lower bound* >= target nu condition.

```python
PRICE = {  # currency per kg  <- REPLACE with real values
    'cement': 0.12, 'slag': 0.05, 'ash': 0.03, 'water': 0.001,
    'superplastic': 1.5, 'coarseagg': 0.02, 'fineagg': 0.02,
}
CO2 = {'cement': 0.9, 'slag': 0.07, 'ash': 0.01, 'water': 0.0,
       'superplastic': 0.25, 'coarseagg': 0.005, 'fineagg': 0.005}  # kg CO2/kg (approx, replace with your data)

def find_cheapest(target_mpa, age_days=28, n=40000, seed=SEED):
    rng = np.random.default_rng(seed)
    lo_b, hi_b = df[MIX_COLS].min(), df[MIX_COLS].max()          # training range-kulla mattum search (extrapolation avoid)
    cand = pd.DataFrame({c: rng.uniform(lo_b[c], hi_b[c], n) for c in MIX_COLS})
    cand['age'] = age_days
    cand = add_features(cand.assign(strength=0)).drop(columns='strength')
    cand = cand[(cand['w_binder'].between(0.25, 0.8))]            # realistic water/binder only
    cand = cand[FEATURES]
    cand['pred']  = best_model.predict(cand)
    cand['lower'] = q_lo.predict(cand[FEATURES])
    ok = cand[cand['lower'] >= target_mpa].copy()
    if ok.empty: return None
    ok['cost'] = sum(ok[c] * PRICE[c] for c in MIX_COLS)          # per m3 (kg values)
    ok['co2']  = sum(ok[c] * CO2[c]   for c in MIX_COLS)
    return ok.sort_values('cost').head(5)[MIX_COLS + ['pred', 'lower', 'cost', 'co2']].round(1)

find_cheapest(40)
```

## Step 10: Save model
Later Streamlit app-la `joblib.load` pannitu use pannalam.

```python
joblib.dump({'model': best_model, 'q_lo': q_lo, 'q_hi': q_hi, 'features': FEATURES}, 'concrete_model.joblib')
print('saved')
```
