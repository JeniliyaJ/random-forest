# Concrete Strength Prediction (Beginner-friendly, professional workflow)

## Big picture (mudhalla idha purinjukkonga)
- **Problem:** Concrete mix (cement, water, age...) therinja, adhoda **strength (MPa)** enna nu predict pannanum.
- **X** = input columns (cement, water, age...). **y** = answer (strength).
- Model-ku **train data** kudutthu kathukka vaikkrom. Apram **test data**-la (adhu paathadhey illa) exam vekkrom.
- Exam score thaan model-oda **real performance**.

## Indha notebook-la steps
1. Data load + clean
2. Konjam EDA
3. New features (domain knowledge)
4. Train/Test split (sariyaana vazhi)
5. 4 models: Linear (basic), Linear (improved), Random Forest, Gradient Boosting
6. Compare + overfitting check
7. Cross-validation (score stable-aa nu check)
8. Feature importance (edhu mukkiyam)
9. Business use: what-if + cheapest mix
10. Save model

## Step 1: Libraries import

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GroupShuffleSplit, GroupKFold, cross_val_score
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

SEED = 42   # random_state: ennaikum same result varum
```

## Step 2: Data load + check

```python
df = pd.read_csv('concrete.csv')
print(df.shape)
df.head()
```

```python
print('Missing values :', df.isna().sum().sum())
print('Duplicate rows :', df.duplicated().sum())
```

### Duplicates remove pannrom
**Yen?** Same row train-lum test-lum irundha, model "answer-ai paathu" mark vaangidum. Adhu fake score.

```python
df = df.drop_duplicates().reset_index(drop=True)
print(df.shape)
```

## Step 3: Quick EDA (data-vai purinjukka)

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation: strength-oda edhu related?')
plt.show()
```

```python
sns.histplot(df['strength'], kde=True)
plt.title('Strength distribution')
plt.show()
```

## Step 4: Feature Engineering (new useful columns)
Concrete science-la theriyum: strength-ku **water kammiya irukkanum**, **age-oda log curve** follow pannum.
Adhanaala:
- `binder` = cement + slag + ash (ellam serndha "paste")
- `w_binder` = water / binder  (romba mukkiyam!)
- `log_age` = log(age)  (age 1->7 romba maarum, 90->365 konjam thaan maarum)

Idhu row-by-row calculation, so leakage prechanai illa.

```python
def add_features(d):
    d = d.copy()
    d['binder']    = d['cement'] + d['slag'] + d['ash']
    d['w_binder']  = d['water'] / d['binder']
    d['w_cement']  = d['water'] / d['cement']
    d['log_age']   = np.log(d['age'])
    return d

df = add_features(df)
df.head()
```

## Step 5: Train / Test split (sariyaana vazhi)

**Prechanai:** Dataset-la oru mix-ai 3, 7, 28, 90 days-la test pannirukkanga. Random split pannina, oru mix-oda 28-day row train-la, 90-day row test-la poidum. Model "andha mix-ai already paathirukken" nu easy-ah guess pannidum. Idhu exam-la **same question paper munnadiye kudutha** madhiri.

**Solution:** Same mix-oda ella rows-um onnaa **train-la mattum** or **test-la mattum** irukkanum. (`GroupShuffleSplit`)
Group = 7 ingredient values (age thavira).

```python
MIX_COLS = ['cement','slag','ash','water','superplastic','coarseagg','fineagg']

# Same ingredients irundha same group
groups = df[MIX_COLS].round(1).astype(str).agg('-'.join, axis=1)
print('Total rows:', len(df), '| Unique mixes:', groups.nunique())

X = df.drop(columns='strength')
y = df['strength']

splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=SEED)
train_idx, test_idx = next(splitter.split(X, y, groups))

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
g_train = groups.iloc[train_idx]
print('Train rows:', len(X_train), '| Test rows:', len(X_test))
```

## Step 6: Metrics (model-ai epdi mark podrom)
- **R²**: 1-ku nerukkama irundha nalladhu. (0.90 = variation-la 90% model explain pannudhu)
- **RMSE**: average error MPa-la (periya mistakes-ku adhiga penalty)
- **MAE**: average error MPa-la (simple) - "average-ah evlo MPa thappu"

Idhu oru helper function, ella models-ukkum use pannalaam.

```python
results = []

def evaluate(name, model, cols):
    # train data-la mattum fit
    model.fit(X_train[cols], y_train)
    # train & test rendulayum predict
    p_train = model.predict(X_train[cols])
    p_test  = model.predict(X_test[cols])
    row = {
        'model': name,
        'Train R2': r2_score(y_train, p_train),
        'Test R2':  r2_score(y_test, p_test),
        'Test RMSE': np.sqrt(mean_squared_error(y_test, p_test)),
        'Test MAE':  mean_absolute_error(y_test, p_test),
    }
    results.append(row)
    return p_test
```

## Step 7: Model 1 - Linear Regression (basic, unga original madhiri)
`make_pipeline(StandardScaler(), LinearRegression())` = "scale pannu, apram model-ku kudu" nu oru box.
Idhu thaan **pipeline**. Fit pannum podhu scaler train data-la mattum kathukkum (leakage illa).

```python
BASIC_COLS = MIX_COLS + ['age']

linear_basic = make_pipeline(StandardScaler(), LinearRegression())
evaluate('Linear (basic)', linear_basic, BASIC_COLS)
pd.DataFrame(results).round(3)
```

## Step 8: Model 2 - Linear (improved features)
Same Linear model, aana `age` badhila `log_age` + `w_binder` kudukrom.
Model maaralla, **features maathi** improve pannrom. Idhu real projects-la romba common.

```python
IMPROVED_COLS = MIX_COLS + ['log_age', 'w_binder']

linear_improved = make_pipeline(StandardScaler(), LinearRegression())
evaluate('Linear (improved)', linear_improved, IMPROVED_COLS)
pd.DataFrame(results).round(3)
```

## Step 9: Model 3 - Random Forest
Pala decision trees serndhu vote pannum. Non-linear relations-ai kathukkum.
`min_samples_leaf=3` = oru leaf-la kurainja 3 rows (romba ezhuthikaama irukka, overfitting kammi).

```python
rf = RandomForestRegressor(n_estimators=300, min_samples_leaf=3, random_state=SEED, n_jobs=-1)
evaluate('Random Forest', rf, IMPROVED_COLS)
pd.DataFrame(results).round(3)
```

## Step 10: Model 4 - Gradient Boosting
Trees oru oru onna build aagum, ovvonnum munnadhi tree-oda mistakes-ai sari pannum.
Indha mathiri tabular data-la bothu RF-ai vida better-ah irukkum.

```python
gb = HistGradientBoostingRegressor(learning_rate=0.05, max_iter=300,
                                   min_samples_leaf=10, random_state=SEED)
evaluate('Gradient Boosting', gb, IMPROVED_COLS)
pd.DataFrame(results).round(3)
```

## Step 11: Compare + overfitting check
**Overfitting** = Train score romba high, Test score kammi. (Model "kathukkala", "manappaadam pannirukku").
`Train R2 - Test R2` gap periya-ah irundha overfitting.

```python
res = pd.DataFrame(results).round(3)
res['Gap (Train-Test R2)'] = (res['Train R2'] - res['Test R2']).round(3)
res
```

```python
res.set_index('model')['Test R2'].plot.barh(figsize=(7,3), title='Test R2 (adhigam = nalladhu)')
plt.show()
```

## Step 12: Cross-validation (score stable-aa?)
Oru test split-la mattum adhirshtam-la score nalla/kettu varalaam. So data-ai 5 pagam pannu, 5 dhadavai train/test panni average edukrom.
Idhula `GroupKFold` use pannrom (Step 5 same reason).

```python
cv = GroupKFold(n_splits=5)

for name, model, cols in [
    ('Linear (basic)',    linear_basic,    BASIC_COLS),
    ('Linear (improved)', linear_improved, IMPROVED_COLS),
    ('Random Forest',     rf,              IMPROVED_COLS),
    ('Gradient Boosting', gb,              IMPROVED_COLS),
]:
    scores = cross_val_score(model, X_train[cols], y_train, groups=g_train, cv=cv, scoring='r2')
    print(f'{name:20s} R2 = {scores.mean():.3f} +- {scores.std():.3f}')
```

**Padikkum podhu:** `0.90 +- 0.02` = average 0.90, 5 folds-la 0.88 - 0.92 range.
`+-` periya-ah irundha model stable illa.

## Step 13: Best model select + test-la visualize
(Results table-la Test R2 highest, gap kammi-ah irukkara model-ai select pannunga. Ippo Gradient Boosting nu assume pannrom. Maathikkalaam.)

```python
best_name = 'Gradient Boosting'
best_model = gb
best_cols = IMPROVED_COLS

pred = best_model.predict(X_test[best_cols])

plt.figure(figsize=(5,5))
plt.scatter(y_test, pred, alpha=0.6)
lim = [y_test.min(), y_test.max()]
plt.plot(lim, lim, 'r--', label='Perfect prediction')
plt.xlabel('Actual strength (MPa)'); plt.ylabel('Predicted strength (MPa)')
plt.title(f'{best_name}: Actual vs Predicted'); plt.legend()
plt.show()
```

```python
errors = y_test - pred
sns.histplot(errors, kde=True)
plt.title('Prediction errors (0 around irundha nalladhu)')
plt.xlabel('Error (MPa)')
plt.show()
print(f'90% predictions +-{np.percentile(np.abs(errors), 90):.1f} MPa kulla')
```

## Step 14: Feature importance - edhu mukkiyam?
**Permutation importance:** oru column-ai random-ah kalachchu pottaa score evlo kuraiyudhu nu paakum.
Evlo kuraiyudho avlo andha column mukkiyam.

```python
from sklearn.inspection import permutation_importance

pi = permutation_importance(best_model, X_test[best_cols], y_test, n_repeats=20,
                            random_state=SEED, scoring='r2')
imp = pd.Series(pi.importances_mean, index=best_cols).sort_values()
imp.plot.barh(figsize=(6,4), title='Feature importance (score drop)')
plt.show()
```

## Step 15: Business use 1 - What-if tool
Mix-ai kudutha strength predict pannum. Site engineer-ku useful.

```python
def predict_strength(cement, slag, ash, water, superplastic, coarseagg, fineagg, age):
    row = pd.DataFrame([{
        'cement': cement, 'slag': slag, 'ash': ash, 'water': water,
        'superplastic': superplastic, 'coarseagg': coarseagg, 'fineagg': fineagg, 'age': age
    }])
    row = add_features(row.assign(strength=0)).drop(columns='strength')
    return float(best_model.predict(row[best_cols])[0])

# Example: same mix, vera vera age
for d in [3, 7, 28, 90]:
    s = predict_strength(300, 100, 0, 180, 6, 950, 780, d)
    print(f'Age {d:3d} days -> {s:.1f} MPa')
```

## Step 16: Business use 2 - Cheapest mix finder
Target strength (e.g. 40 MPa) kidaikkara **cheapest** mix-ai search pannum.

- **PRICE: placeholder.** Unga company real price (per kg) podunga.
- **Safety margin:** Model-ku error irukku, so `target + margin` predict aaganum. Margin = test RMSE.
- Training data range-kulla mattum search pannrom (model range-ku veliya nambakoodathu).

```python
PRICE = {'cement': 0.12, 'slag': 0.05, 'ash': 0.03, 'water': 0.001,
         'superplastic': 1.5, 'coarseagg': 0.02, 'fineagg': 0.02}   # <- REPLACE

def find_cheapest(target_mpa, age_days=28, n_samples=30000):
    rng = np.random.default_rng(SEED)
    margin = np.sqrt(mean_squared_error(y_test, pred))     # safety margin = test RMSE

    # random mixes generate (training range kulla)
    cand = pd.DataFrame({c: rng.uniform(df[c].min(), df[c].max(), n_samples) for c in MIX_COLS})
    cand['age'] = age_days
    cand = add_features(cand.assign(strength=0)).drop(columns='strength')
    cand = cand[cand['w_binder'].between(0.25, 0.8)]       # realistic water/binder

    cand['predicted'] = best_model.predict(cand[best_cols])
    ok = cand[cand['predicted'] >= target_mpa + margin].copy()
    if ok.empty:
        return 'Indha target-ku mix kidaikkala'
    ok['cost_per_m3'] = sum(ok[c] * PRICE[c] for c in MIX_COLS)
    return ok.sort_values('cost_per_m3').head(5)[MIX_COLS + ['predicted', 'cost_per_m3']].round(1)

find_cheapest(target_mpa=40)
```

## Step 17: Model save (later app-la use pannalam)

```python
joblib.dump({'model': best_model, 'columns': best_cols}, 'concrete_model.joblib')
print('Saved: concrete_model.joblib')
```
